```python

# Praktikum Algoritma dan Pemrograman 1 - Variable dan Tipe Data

# Tipe data Boolean
benar = False
print(benar)
print(type(benar))

# Tipe data String
teks = "Ini adalah tulisan berupa String"
print(teks)

# Tipe data Integer
angka = 100
print(angka)

# Tipe data Float
desimal = 0.001
print(desimal)

# Tipe data Hexadecimal
heksa = 0x01
print("bilangan desimal dari 0x01 adalah", heksa)

# Tipe data Octal
oktal = 0o12
print(oktal)

# Tipe data Complex
kompleks = 2 + 6j
print(type(kompleks))
print(kompleks)

# Tipe data Complex lainnya
kompleks2 = 3 + 4j
print(type(kompleks2))

# Tipe data List
angka_list = [96, 97, 98, 99, 100]
print(angka_list)

# List String
huruf = ['seratus', 'dua ratus', 'tiga ratus']
print(huruf)

# Tipe data Dictionary
biodata = {'nama': 'Ani', 'umur': 19}
print(biodata)

# String dengan tanda kutip satu di dalamnya
kalimat = "This string contains a single quote (') character."
print(kalimat)
```
